```python
import pandas as pd
```

```python
input_files = ["Thông tin chim_bổ sung.csv","Thông tin lưỡng cư bò sát_bổ sung.csv","Thông tin thú_bổ sung.csv"]
dest_file = ["Thông tin chim_bổ sung.csv","Thông tin lưỡng cư bò sát_bổ sung.csv","Thông tin thú_bổ sung.csv"]
```

```python
for i, file in enumerate(input_files):
    print(i, file)
```

```python
for i, file in enumerate(input_files):
    print(file)
    df = pd.read_csv(file, sep = "|")
    df = df.fillna("")
    df.to_csv(dest_file[i], sep="|", index=False)
```

```python
df = pd.read_csv("Thông tin chim_bổ sung.csv", sep="|")
```

```python
df = pd.read_excel(input_files[0])
```

```python
# Xoá các cột có tên bắt đầu bằng 'Unnamed'
df = df.loc[:, ~df.columns.str.startswith('Unnamed')]

df.info()
```

```python
for col in df.columns:
    df[col] = df[col].map(lambda x: str(x).replace('\n', '').replace('\r', ''))
```

```python
df_2 = pd.read_csv(input_files[1], sep="|")
```

```python
df = df.merge(df_2[['Tên khoa học', 'Vùng phân bố']], on='Tên khoa học', how='left')
```

```python
df['Vùng phân bố'] = df['Vùng phân bố'].str.replace('|', ',', regex=False)
```

```python
df.info()
```

```python
df.to_csv(dest_file, sep="|", index=False)
```
